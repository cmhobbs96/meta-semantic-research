```python
import sys
import os
import pandas as pd

sys.path.append(os.path.abspath("../"))

from scripts.util.getters import get_results_path

enhancements = ["raw", "srl", "kga", "rbm", "srt"]
test_rows = []
gen_rows = []

for enh in enhancements:
    for dtype, rows in [("test", test_rows), ("gen", gen_rows)]:
        path = get_results_path("t5-small", dtype, enh)
        try:
            df = pd.read_csv(path)
            df["enhancement"] = enh
            rows.append(df)
        except FileNotFoundError:
            print(f"[!] Missing: {path}")

# Combine and reorder columns
test_df = pd.concat(test_rows, ignore_index=True)
gen_df = pd.concat(gen_rows, ignore_index=True)

cols = ["enhancement"] + [col for col in test_df.columns if col != "enhancement"]
test_df = test_df[cols]
gen_df = gen_df[cols]

print("\n--- Test Results ---")
print(test_df.to_markdown(index=False))

print("\n--- Gen Results ---")
print(gen_df.to_markdown(index=False))
```
