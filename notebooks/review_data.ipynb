```python
# --- Imports ---
import pandas as pd
from IPython.display import display
import sys
import os

# Add root directory to path
sys.path.append(os.path.abspath(".."))

from scripts.util.getters import get_dataset_paths
```

```python
# --- Config ---
ENHANCEMENTS = ["raw", "srl", "kga", "rbm", "srt"]
```

```python
# --- Helper Function ---
def review_dataset(path: str, name: str):
    print(f"\nReviewing {name} — ({path})")

    try:
        df = pd.read_csv(path, sep="\t", header=None)
    except FileNotFoundError:
        print(f"File not found: {path}")
        return

    print("Loaded:", name)
    print("Shape:", df.shape)
    display(df.head(3))

    if df.shape[1] >= 2:
        print("Input Distribution:")
        print(df[0].value_counts().head(), "\n")

        print("Output Distribution:")
        print(df[1].value_counts().head(), "\n")

    if df.shape[1] == 3:
        print("Roles Distribution:")
        print(df[2].value_counts().head(), "\n")
```

```python
paths = get_dataset_paths("raw")
for split in ["train", "dev", "test", "gen"]:
    review_dataset(paths[split], name=f"RAW - {split}")
```

```python
paths = get_dataset_paths("srl")
review_dataset(paths["train"], name=f"SRL - train")
```

```python
paths = get_dataset_paths("srt")
review_dataset(paths["train"], name=f"SRT - train")
```
